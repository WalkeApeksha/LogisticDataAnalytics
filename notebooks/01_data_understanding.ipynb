```python
import pandas as pd
import os

file_path = "../data/raw/routes_performance.xlsx"

print("File exists:", os.path.exists(file_path))

# Show all sheets in the Excel file
excel_file = pd.ExcelFile(file_path)

print("Sheet names:")
print(excel_file.sheet_names)
```

```python
# Load the first sheet
df = pd.read_excel(file_path, sheet_name=excel_file.sheet_names[0])

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("first 5 rows")
print(df.head())
```

```python
print("data types")
print(df.dtypes)
```

```python
print("MIssing values")
print(df.isnull().sum())
```

```python
print("Duplicated rows")
print(df.duplicated().sum())
```

```python
print("Basis Statistics : ")
print(df.describe(include="all"))
```

```python
print("Column names:")
for col in df.columns:
    print("-",col)
```

```python
print("unique values")
for col in df.columns:
    print("\n",col)
    print(df[col].nunique())
```

```python
df["Route Deviation"]=df["DistanceA"]-df["DistanceP"]
print("Route Deviation Calculated successfully:")
display(df[["DistanceA","DistanceP","Route Deviation"]].head())
```

```python
print(" Average planned Distance : ",df["DistanceP"].mean())
print("Average Actual Distance: ",df["DistanceA"].mean())
print("Average Route Deviation: ",df["Route Deviation"].mean())
```

```python
# Check whether delivery arrived within the allowed time window

df["Within Time Window"] = (
    (df["Arrived Time"] >= df["Earliest Time"]) &
    (df["Arrived Time"] <= df["Latest Time"])
)

print(df["Within Time Window"].value_counts())
```

```python
time_window_compliance = df["Within Time Window"].mean() * 100

print(
    f"Time-Window Compliance: {time_window_compliance:.2f}%"
)
```

```python
import matplotlib.pyplot as plt

labels = ["Planned Distance", "Actual Distance"]
values = [
    df["DistanceP"].mean(),
    df["DistanceA"].mean()
]

plt.bar(labels, values)
plt.title("Average Planned vs Actual Distance")
plt.ylabel("Distance")
plt.show()
```
